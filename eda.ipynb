```python
titanics.head()
```

```python
sns.barplot(x='Pclass',y='Age',data=titanics,hue='Sex')
```

```python
sns.boxplot(x='Sex',y='Age',data=titanics,hue='Survived')
```

```python
sns.kdeplot(x='Age',data=titanics[titanics['Survived']==0],fill=False)
sns.kdeplot(x='Age',data=titanics[titanics['Survived']==1],fill=False)
```

```python
sns.heatmap(pd.crosstab(titanics['Pclass'],titanics['Survived']))
```

```python
from seaborn.matrix import clustermap

sns.clustermap = clustermap

sns.clustermap(pd.crosstab(titanics['Parch'], titanics['Survived']))
plt.show()
```
